# BidirectionalRNN

```python
import tensorflow as tf 
import keras 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Embedding,SimpleRNN ,LSTM,GRU
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
# Load the dataset

num_words =100000  # number of unique words to keep


(X_train ,Y_train) ,(X_test,Y_test ) = imdb.load_data(num_words=num_words) 

# pad_sequences

maxlen = 100 # pad_sequences length
X_train =pad_sequences(X_train,maxlen=maxlen ,padding ='post',truncating='post')
X_test =pad_sequences(X_test,maxlen=maxlen ,padding='post',truncating='post')

```

```python
# Simple RNN Architecture

embedding_dim =32  # Dimension of embedding layer

model =Sequential([
    Embedding(input_dim=num_words ,output_dim=embedding_dim ,input_length =maxlen) ,
    SimpleRNN(5) ,
    Dense(1,activation='sigmoid')
    
])
model.build(input_shape=(None,100))
model.compile(loss ='binary_crossentropy',optimizer ='adam',metrics =['accuracy'])


model.summary() # notice the weights count
```

```python
from tensorflow.keras.layers import Bidirectional

# Building the Biderictional RNN model
embedding_dim =32  # Dimension of embedding layer

model =Sequential([
    Embedding(input_dim=num_words ,output_dim=embedding_dim ,input_length =maxlen) ,
    Bidirectional(SimpleRNN(5)) ,  # the only change Bidirectional
    Dense(1,activation='sigmoid')
])

model.build(input_shape=(None,100))

model.compile(loss ='binary_crossentropy',optimizer ='adam',metrics =['accuracy'])

model.summary()
```
