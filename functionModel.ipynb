```python
x =Input(shape =(3,))
```

# Mutliple ouput model

```python
# Define hidden layers
hidden1 = Dense(128, activation='relu')(x)
hidden2 = Dense(64, activation='relu')(hidden1)

# Define multiple outputs
output1 = Dense(1, activation='linear')(hidden2)  # Regression output
output2 = Dense(1, activation='sigmoid')(hidden2) # Classification output

# Define the model with multiple outputs
model = Model(inputs=x, outputs=[output1, output2])

# Display model architecture
model.summary()
```

```python
import pydot
import graphviz
from keras.utils import plot_model
plot_model(model,show_shapes =True)
```

# Multiple Input  model

```python
inputA =Input(shape=(32,))
inputB =Input(shape=(128,))
```

```python
# the first branch operator on the first ouput
x =Dense(8,activation='relu')(inputA)
x1 =Dense(4,activation='relu')(x)
```

```python
# the second branch operator on the first ouput
y =Dense(8,activation='relu')(inputB)
y1 =Dense(4,activation='relu')(y)
y2 =Dense(4,activation='relu')(y1)
```

```python
import pandas as pd  
import numpy as np
combine =concatenate([x1,y2])
```

```python
# Apply a fullyconnected layer and then a regression predication
z =Dense(2,activation='relu')(combine)
z1 =Dense(1,activation='linear')(z)
```

```python
model =Model(inputs =[inputA,inputB] ,outputs =z1)
```

```python
model.summary()
```

```python
from keras.utils import plot_model
plot_model(model,show_shapes =True)
```
