# Whatsapp chat analyzer


```python
import pandas as pd 
import re # for regularexpression
```

```python
f = open("WhatsApp Chat with It Unofficial 2022.txt",'r', encoding='utf-8')  #UTF-8 (Unicode Transformation Format - 8 bit)



#UTF-8 is a character encoding standard that allows computers to store and exchange text in multiple languages and symbols efficiently. It is the most widely used encoding on the web.
```

```python
data = f.read()
```

```python
print(data) 
```

```python
print(type(data))
```

```python
# regular expression for identfiy the pattern 
# go on regex101.com print a sample message and got your regular expression


actual_message = r"^(\d{1,2}/\d{1,2}/\d{2}), (\d{1,2}:\d{2})\s?([AP]M) - ([A-Za-z\s]+): (.+)$"

pattern  = "\d{1,2}/\d{1,2}/\d{2,4},\s\d{1,2}:\d{2}\s?[AP]M\s-\s"
time_slot ="?[AP]M\s-\s"
```

```python
message =re.split(pattern,data)[1:] # [1:] as empty list is apperaing in it

message
```

```python

dates =re.findall(pattern,data)
dates = [i.replace("\u202f"," ") for i in dates]
dates = [i.replace("-","") for i in dates]
dates
 
```

```python
df = pd.DataFrame({"user_message": message,"date":dates})

# convert message_date type
df['date'] = pd.to_datetime(df['date'].str.strip() ,format = '%m/%d/%y, %I:%M %p')
df.head()
```

```python
# separate user and message 
users = []
messages =[]
for message in df['user_message']:
    entry = re.split('([\w\W]+?):\s',message)
    if entry [1:]: # user message
        users.append(entry[1])
        messages.append(entry[2])
    else:
        users.append('group_notification')
        messages.append(entry[0]) 
        
df['user'] =users
df['messages'] =messages
df.drop(columns=['user_message'],inplace=True) 

df.sample(10)  
```

```python
df['Day'] = df['date'].dt.day_name()
df['Month'] = df['date'].dt.month_name()

df['Year'] = df['date'].dt.year
df.drop(columns=['date'],inplace=True)
```
